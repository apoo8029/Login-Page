```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# 1. Model aur Tokenizer load karein
model_name = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# 2. Apna paragraph
text = """Artificial Intelligence has seen massive growth over the past few years,
revolutionizing industries from healthcare to finance. Modern deep learning architectures,
such as Transformers, allow models to process vast amounts of unstructured data like text,
audio, and video with unprecedented accuracy. As open-source communities like Hugging Face
continue to expand, researchers and developers worldwide can collaborate, share pre-trained
weights, and build intelligent applications without needing massive supercomputers from scratch."""

# 3. Text ko summarize karein
inputs = tokenizer([text], max_length=1024, return_tensors="pt", truncation=True)
summary_ids = model.generate(inputs["input_ids"], max_length=50, min_length=20, do_sample=False)

# 4. Output print karein
print("Summary:", tokenizer.decode(summary_ids[0], skip_special_tokens=True))
```

```python
# Ab hum koi bhi naya text de sakte hain aur max_length badha sakte hain
new_text = """Machine learning interviews often test a candidate's practical understanding of
how models are deployed in production rather than just theoretical formulas. Knowing how to load
pre-trained models using Hugging Face Transformers, tokenize raw text inputs, and generate
predictions in Python gives students a strong competitive edge during campus placements and hackathons."""

inputs = tokenizer([new_text], max_length=1024, return_tensors="pt", truncation=True)

# max_length ko 80 kar diya taaki sentence adhoora na kate
summary_ids = model.generate(inputs["input_ids"], max_length=80, min_length=25, do_sample=False)

print("Complete Summary:", tokenizer.decode(summary_ids[0], skip_special_tokens=True))
```
